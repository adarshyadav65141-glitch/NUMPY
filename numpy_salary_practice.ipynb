```python
import numpy as np
# Columns: Salary | Experience (Years) | Performance Score
salary = np.array([ [25000, 2, 80], 
                    [45000, 5, 90], 
                    [30000, 3, 75], 
                    [60000, 8, 95],
                    [35000, 4, 85] ])
# 1. How will you calculate the average salary of all employees?
avg_salary=np.average(salary[:,0])
print(avg_salary)
```

```python
# 2. How will you find the highest salary?
highest_sal=np.max(salary[:,0])
print(highest_sal)
```

```python
# 3. How will you find the lowest salary?
lowest_sal=np.min(salary[:,0])
print(lowest_sal)
```

```python
# 4. How will you calculate the average experience?
avg_exprice=np.average(salary[:,1])
print(avg_exprice)
                
```

```python
# 5. How will you find employees whose salary is greater than ₹40,000?
salary_greater=salary[salary[:,0]>40000]
print(salary_greater)
```

```python
# 6. How will you find employees whose performance score is greater than 80?
performance_score=salary[salary[:,2]>80]
print(performance_score)
```

```python
# 7. How will you find the employee with the highest performance score?
high_performance=salary[salary[:,2]==np.max(salary[:,2])]
print(high_performance)
```

```python
# 8. How will you calculate the standard deviation of salaries?
standard_deviation =np.std(salary[:,0])
print(standard_deviation)
```

```python
# 9. How will you classify employees as High Salary or Low Salary using np.where()?
salary_level = np.where(salary[:, 0] >= 40000, "High Salary", "Low Salary")
print(salary_level)
```

```python
# 10. How will you convert the results into a Pandas DataFrame?
import pandas as pd

df = pd.DataFrame(
    salary,
    columns=["Salary", "Experience", "Performance"]
)

df["Salary_Level"] = salary_level

print(df)
```
