```python
import numpy as np
attendance = np.array([
 [90, 85, 95, 88],
 [75, 80, 70, 78],
 [95, 92, 98, 96],
 [65, 70, 72, 68],
 [85, 88, 90, 87]
])
# Columns: Python | Java | SQL | ML
# 1. How will you calculate the average attendance of each student?
avg_attendance=np.average(attendance,axis=1)
print(avg_attendance)

```

```python
# 2. How will you calculate the average attendance of each subject?
avg_subject=np.average(attendance,axis=0)
print(avg_subject)
```

```python
# 3. How will you find the highest attendance in each subject?
print(np.max(attendance,axis=0))
```

```python
# 4. How will you find the lowest attendance in each subject?
print(np.min(attendance,axis=0))
```

```python
# 5. How will you find students having average attendance above 80%?

result=attendance[avg_attendance>80]
print(result)
```

```python
# 6. How will you find the student with the highest average attendance?
hightest_avg=np.max(avg_attendance)
print(hightest_avg)
```

```python
# 7. How will you calculate the standard deviation of attendance?
standard_deviation=np.std(attendance)
print(standard_deviation)
```

```python
# 8. How will you use np.where() to classify students as Eligible or Not Eligible based on 75% attendance?
students=np.where(avg_attendance>75,"Eligible","Not Eligible")
print(students)

```

```python
# 9. How will you convert the data into a Pandas DataFrame?
import pandas as pd
df=pd.DataFrame(attendance,
              columns=["Python","Java"," SQL"," ML"]
            )
print(df)
```

```python
# 10. How will you add a Status column to the DataFrame?
# df=pd.DataFrame(attendance
df["Statuts"]=students
print(df)
```
