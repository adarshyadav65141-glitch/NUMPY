```python
import numpy as np
sales = np.array([
 [12000, 15000, 18000],
 [10000, 14000, 16000],
 [18000, 20000, 22000],
 [9000, 12000, 15000],
 [15000, 17000, 19000]
])
# Columns: January | February | March
# 1. How will you calculate the total sales of each salesperson?
total_sale=np.sum(sales,axis=1)
print(total_sale)
```

```python
# 2. How will you calculate the average monthly sales for each salesperson?
avg_monthly=np.average(sales,axis=1)
print(avg_monthly)
```

```python
# 3. How will you find the highest sales achieved in each month?
highest_sale=np.max(sales,axis=0)
print(highest_sale)
```

```python
# 4. How will you find the lowest sales achieved in each month?
print(np.min(sales,axis=0))
```

```python
# 5. How will you find the salesperson with the highest total sales?
hight_total_sale=np.max(total_sale)
print(hight_total_sale)
```

```python
# 6. How will you find salespersons whose average sales are above ₹15,000?
avgerage_sale=np.average(sales,axis=1)
# print(avgerage_sale)
avg_sale=sales[avgerage_sale>15000]
print(avg_sale)
```

```python
# 7. How will you calculate the total company sales for each month?
total_company=np.sum(sales,axis=0)
print(total_company)
```

```python
# 8. How will you calculate the standard deviation of monthly sales?
standard_deviation=np.std(sales,axis=0)
print(standard_deviation)
```

```python
# 9. How will you classify salespersons as Excellent, Good or Needs Improvement using np.where()?
salespersion=np.where(sales>=18000,"Excellent",
             np.where(sales>=15000,"Good",
                       "Needs Improvement"
                     )
                     )
print(salespersion)
```

```python
# 10. How will you convert the final results into a Pandas DataFrame?
import pandas as pd
df=pd.DataFrame(
    sales,
    columns=[ "January","February","March"]
)
print(df)
```
